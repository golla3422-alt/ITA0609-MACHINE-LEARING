```python

# Question 6: Naive Bayes Algorithm
# Dataset 3: Disease Diagnosis

# Step 1: Import required libraries
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score

# Step 2: Create the dataset
X = [
    [39.1, 112, 94],
    [38.8, 109, 95],
    [37.0, 78, 99],
    [39.3, 115, 93],
    [36.8, 74, 98],
    [38.7, 110, 94],
    [37.1, 80, 98],
    [39.0, 113, 92],
    [36.9, 76, 99],
    [38.9, 111, 93]
]

y = [
    "Positive",
    "Positive",
    "Negative",
    "Positive",
    "Negative",
    "Positive",
    "Negative",
    "Positive",
    "Negative",
    "Positive"
]

# Step 3: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 4: Train the Naive Bayes model
model = GaussianNB()
model.fit(X_train, y_train)

# Step 5: Predict class labels
y_pred = model.predict(X_test)

# Step 6: Calculate confusion matrix and accuracy
cm = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)

# Step 7: Predict a new sample
new_sample = [[38.7, 109, 94]]
prediction = model.predict(new_sample)

# Step 8: Display results
print("Confusion Matrix:")
print(cm)

print("\nActual Values:", y_test)
print("Predicted Values:", y_pred)

print("\nAccuracy:", accuracy * 100, "%")

print("\nNew Sample:")
print("Temperature: 38.7")
print("Heart Rate: 109")
print("Oxygen Level: 94")

print("Predicted Disease Class:", prediction[0])
```
