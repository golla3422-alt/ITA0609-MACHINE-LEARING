```python

# EXPERIMENT 1: FIND-S ALGORITHM
# Dataset 1: Play Tennis

import pandas as pd

# Step 1: Create the dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm', 'Warm', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High', 'Normal', 'Normal'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong', 'Weak', 'Weak'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool', 'Warm', 'Warm'],
    'Forecast': ['Same', 'Same', 'Change', 'Same', 'Same', 'Same'],
    'PlayTennis': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Step 2: Display the dataset
print("TRAINING DATASET:")
display(df)

# Step 3: Implement FIND-S algorithm
attributes = ['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast']

hypothesis = ['Ø'] * len(attributes)

# Step 4: Consider only positive examples
for index, row in df.iterrows():
    if row['PlayTennis'] == 'Yes':
        example = [row[attr] for attr in attributes]

        if hypothesis[0] == 'Ø':
            hypothesis = example.copy()
        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != example[i]:
                    hypothesis[i] = '?'

        print("Positive Example:", example)
        print("Updated Hypothesis:", hypothesis)
        print()

# Step 5: Display final hypothesis
print("FINAL SPECIFIC HYPOTHESIS:")
print(hypothesis)
```

```python

# Save the dataset as a CSV file

df.to_csv('play_tennis.csv', index=False)

print("Dataset saved successfully!")

# Download the CSV file
from google.colab import files
files.download('play_tennis.csv')
```
