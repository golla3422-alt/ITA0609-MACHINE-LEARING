```python

from sklearn.mixture import GaussianMixture
import numpy as np

# Step 1: Create the dataset
X = np.array([
    [40, 4.5],
    [50, 5.0],
    [60, 5.6],
    [70, 6.5],
    [80, 7.8],
    [90, 9.2],
    [100, 10.8],
    [110, 12.5],
    [120, 14.3],
    [130, 16.2]
])

# Step 2: Create and train the EM model
gmm = GaussianMixture(
    n_components=2,
    random_state=42
)

gmm.fit(X)

# Step 3: Predict cluster labels
clusters = gmm.predict(X)

# Step 4: Predict cluster for the new sample
new_sample = np.array([[95, 10.0]])
new_cluster = gmm.predict(new_sample)

# Step 5: Display results
print("Vehicle Speed Analysis - EM Clustering")
print("\nDataset:")
print(X)

print("\nCluster Labels:")
print(clusters)

print("\nNew Sample:")
print("Speed = 95 km/h")
print("Fuel Consumption = 10.0 L/100 km")

print("\nNew Sample Cluster:")
print(new_cluster[0])
```
