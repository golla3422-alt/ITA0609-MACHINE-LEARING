```python

import pandas as pd
from sklearn.linear_model import Perceptron

# Create Dataset 2
data = {
    "CGPA": [9.3, 8.9, 8.6, 7.3, 6.9, 9.1],
    "Aptitude": [92, 87, 84, 74, 69, 91],
    "Communication": [90, 85, 82, 72, 67, 89],
    "Placement": [
        "Placed", "Placed", "Placed",
        "Not Placed", "Not Placed", "Placed"
    ]
}

df = pd.DataFrame(data)

# Separate input and output
X = df[["CGPA", "Aptitude", "Communication"]]
y = df["Placement"]

# Create and train Perceptron model
model = Perceptron(max_iter=1000, random_state=42)
model.fit(X, y)

# New student sample
new_sample = pd.DataFrame({
    "CGPA": [8.8],
    "Aptitude": [86],
    "Communication": [84]
})

# Predict placement
prediction = model.predict(new_sample)

print("New Student Details:")
print(new_sample)

print("\nPredicted Placement:")
print(prediction[0])
```
