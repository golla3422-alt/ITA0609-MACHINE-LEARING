```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Create Dataset 5: Automobile Showroom Sales
data = {
    "Marketing Budget": [30, 35, 40, 45, 50, 55, 60, 65, 70, 75],
    "Test Drives": [20, 25, 30, 35, 40, 45, 50, 55, 60, 65],
    "Cars Sold": [8, 10, 13, 17, 22, 28, 35, 43, 52, 62]
}

df = pd.DataFrame(data)

# Input features and target
X = df[["Marketing Budget", "Test Drives"]]
y = df["Cars Sold"]

# Create and train Linear Regression model
model = LinearRegression()
model.fit(X, y)

# New sample for prediction
new_sample = pd.DataFrame({
    "Marketing Budget": [58],
    "Test Drives": [48]
})

# Predict future sales
prediction = model.predict(new_sample)

print("Automobile Showroom Sales Prediction")
print("------------------------------------")
print("New Marketing Budget: 58 (₹ Thousands)")
print("New Test Drives: 48")
print("\nPredicted Future Sales:")
print(round(prediction[0], 2), "Cars")
```
