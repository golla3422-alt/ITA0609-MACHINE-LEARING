```python

# Question 5: K-Nearest Neighbours (KNN)
# Dataset 2: Student Placement Classification

# Step 1: Import required libraries
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Step 2: Create the dataset
X = [
    [9.2, 90, 88],
    [8.8, 85, 82],
    [8.5, 80, 79],
    [7.2, 72, 70],
    [6.8, 68, 65],
    [9.0, 92, 90],
    [7.5, 75, 74],
    [8.6, 84, 81],
    [6.5, 60, 58],
    [8.9, 88, 86]
]

y = [
    "Placed",
    "Placed",
    "Placed",
    "Not Placed",
    "Not Placed",
    "Placed",
    "Not Placed",
    "Placed",
    "Not Placed",
    "Placed"
]

# Step 3: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 4: Train the KNN model (K = 3)
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# Step 5: Predict the class of a new sample
new_sample = [[8.7, 86, 84]]
prediction = knn.predict(new_sample)

# Step 6: Display the results
y_pred = knn.predict(X_test)

print("New Student Details:")
print("CGPA: 8.7")
print("Aptitude Score: 86")
print("Communication Score: 84")

print("\nPredicted Placement:", prediction[0])
print("Model Accuracy:", accuracy_score(y_test, y_pred) * 100, "%")
```
