```python

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Create Dataset 1: Customer Loan Profile
data = {
    "Annual Income": [12, 10, 8, 6, 4, 15, 7, 5, 11, 3],
    "Credit History": [10, 8, 6, 4, 2, 12, 5, 3, 9, 1],
    "Existing Loans": [1, 2, 2, 3, 4, 1, 3, 4, 2, 5],
    "EMI Paid on Time": [
        "Yes", "Yes", "Yes", "No", "No",
        "Yes", "Yes", "No", "Yes", "No"
    ],
    "Credit Score": [
        "Excellent", "Good", "Good", "Average", "Poor",
        "Excellent", "Average", "Poor", "Good", "Poor"
    ]
}

df = pd.DataFrame(data)

# Separate input features and target
X = df.drop("Credit Score", axis=1)
y = df["Credit Score"]

# Encode categorical input column
encoder = LabelEncoder()
X["EMI Paid on Time"] = encoder.fit_transform(
    X["EMI Paid on Time"]
)

# Create and train Decision Tree model
model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

model.fit(X, y)

# New Sample from the lab document
new_sample = pd.DataFrame({
    "Annual Income": [9],
    "Credit History": [7],
    "Existing Loans": [2],
    "EMI Paid on Time": [
        encoder.transform(["Yes"])[0]
    ]
})

# Predict credit score
prediction = model.predict(new_sample)

print("CUSTOMER LOAN PROFILE")
print("---------------------")
print("Annual Income: 9 Lakhs")
print("Credit History: 7 Years")
print("Existing Loans: 2")
print("EMI Paid on Time: Yes")
print("---------------------")
print("Predicted Credit Score:", prediction[0])
```
