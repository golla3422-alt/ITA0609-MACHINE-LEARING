```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Create Dataset 5
data = {
    "Antutu Score": [250000, 320000, 420000, 520000, 620000,
                     720000, 820000, 920000, 1020000, 1120000],
    "Battery": [4000, 4500, 5000, 5000, 5500,
                6000, 6000, 6500, 7000, 7000],
    "Camera": [48, 50, 64, 64, 108, 108, 200, 200, 200, 200],
    "Price": [15000, 18000, 24000, 30000, 38000,
              46000, 58000, 70000, 85000, 100000]
}

df = pd.DataFrame(data)

# Separate input features and target
X = df[["Antutu Score", "Battery", "Camera"]]
y = df["Price"]

# Create and train Linear Regression model
model = LinearRegression()
model.fit(X, y)

# New mobile sample
new_sample = pd.DataFrame({
    "Antutu Score": [750000],
    "Battery": [6000],
    "Camera": [108]
})

# Predict mobile price
prediction = model.predict(new_sample)

print("Mobile Specifications:")
print(new_sample)

print("\nPredicted Mobile Price:")
print("₹", round(prediction[0], 2))
```
