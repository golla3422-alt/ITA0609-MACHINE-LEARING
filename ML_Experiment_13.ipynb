```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
data = {
    "Manufacturing Year": [2024, 2023, 2022, 2021, 2020,
                           2019, 2018, 2017, 2016, 2015],
    "Mileage": [10000, 18000, 25000, 35000, 45000,
                55000, 65000, 75000, 85000, 95000],
    "Fuel Efficiency": [22, 21, 20, 19, 18, 17, 16, 15, 14, 13],
    "Price": [10.5, 9.8, 9.0, 8.2, 7.5, 6.8, 6.0, 5.4, 4.8, 4.2]
}

df = pd.DataFrame(data)

# Separate input and target
X = df[["Manufacturing Year", "Mileage", "Fuel Efficiency"]]
y = df["Price"]

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# New sample
new_sample = pd.DataFrame({
    "Manufacturing Year": [2021],
    "Mileage": [32000],
    "Fuel Efficiency": [19]
})

# Predict car price
prediction = model.predict(new_sample)

print("New Car Details:")
print(new_sample)

print("\nPredicted Car Price:")
print(round(prediction[0], 2), "Lakhs")
```
