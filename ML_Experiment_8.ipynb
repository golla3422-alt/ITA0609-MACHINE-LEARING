```python

# Question 8: Linear Regression
# Dataset 4: Car Fuel Efficiency Prediction

# Step 1: Import the required library
from sklearn.linear_model import LinearRegression

# Step 2: Create the dataset
# Features: Engine Capacity, Vehicle Weight, Speed

X = [
    [1.0, 900, 60],
    [1.2, 950, 70],
    [1.4, 1050, 80],
    [1.6, 1150, 90],
    [1.8, 1250, 100],
    [2.0, 1350, 110],
    [2.2, 1450, 120],
    [2.4, 1550, 130],
    [2.6, 1650, 140],
    [2.8, 1750, 150]
]

# Target: Mileage (km/L)
y = [24, 22, 20, 18, 16, 15, 13, 12, 11, 10]

# Step 3: Create and train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Step 4: Predict mileage for a new sample
new_sample = [[1.5, 1100, 85]]
prediction = model.predict(new_sample)

# Step 5: Display the results
print("Car Details:")
print("Engine Capacity: 1.5 L")
print("Vehicle Weight: 1100 kg")
print("Speed: 85 km/h")

print("\nPredicted Mileage:", round(prediction[0], 2), "km/L")
```
