```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Dataset 1: Student Placement Prediction
data = {
    "CGPA": [9.2, 8.8, 8.5, 7.2, 6.8, 9.0, 7.5, 8.6, 6.5],
    "Aptitude": [90, 85, 80, 72, 68, 92, 75, 84, 60],
    "Communication": [88, 82, 79, 70, 65, 90, 74, 81, 58],
    "Placement": [
        "Placed", "Placed", "Placed",
        "Not Placed", "Not Placed", "Placed",
        "Not Placed", "Placed", "Not Placed"
    ]
}

df = pd.DataFrame(data)

# Separate input and output
X = df[["CGPA", "Aptitude", "Communication"]]
y = df["Placement"]

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Decision Tree model
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)

# KNN model
knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train, y_train)
knn_pred = knn_model.predict(X_test)

# Calculate accuracy
dt_accuracy = accuracy_score(y_test, dt_pred)
knn_accuracy = accuracy_score(y_test, knn_pred)

# Display results
print("Decision Tree Accuracy:", round(dt_accuracy * 100, 2), "%")
print("KNN Accuracy:", round(knn_accuracy * 100, 2), "%")

if dt_accuracy > knn_accuracy:
    print("Decision Tree performed better.")
elif knn_accuracy > dt_accuracy:
    print("KNN performed better.")
else:
    print("Both algorithms achieved the same accuracy.")
```
