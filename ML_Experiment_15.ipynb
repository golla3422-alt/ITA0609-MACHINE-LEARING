```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
data = {
    "House Area": [1000, 1200, 1400, 1600, 1800,
                   2000, 2200, 2400, 2600, 2800],
    "Floors": [1, 1, 2, 2, 2, 3, 3, 3, 4, 4],
    "Garden Area": [100, 120, 150, 180, 200,
                    240, 280, 320, 350, 400],
    "House Price": [45, 55, 68, 82, 96,
                    112, 130, 150, 172, 196]
}

df = pd.DataFrame(data)

# Separate input and target
X = df[["House Area", "Floors", "Garden Area"]]
y = df["House Price"]

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# New sample
new_sample = pd.DataFrame({
    "House Area": [2100],
    "Floors": [3],
    "Garden Area": [250]
})

# Predict house price
prediction = model.predict(new_sample)

print("New House Details:")
print(new_sample)

print("\nPredicted House Price:")
print(round(prediction[0], 2), "Lakhs")
```
